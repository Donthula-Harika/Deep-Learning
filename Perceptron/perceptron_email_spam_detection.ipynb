```python
import numpy as np

# Training data
# X = [x1, x2]
X = np.array([
    [1, 1],  # E1
    [1, 0],  # E2
    [0, 1],  # E3
    [0, 0]   # E4
])

# Labels (Spam = 1, Not Spam = 0)
y = np.array([1, 1, 1, 0])

# Initialize weights and bias
w = np.array([0.0, 0.0])
b = 0.0
lr = 1

# Step activation function
def step(z):
    return 1 if z > 0 else 0

# One full epoch
for i in range(len(X)):
    z = np.dot(w, X[i]) + b
    y_hat = step(z)

    # Perceptron update rule
    w = w + lr * (y[i] - y_hat) * X[i]
    b = b + lr * (y[i] - y_hat)

    print(f"After E{i+1}: w={w}, b={b}")

print("\nFinal weights:", w)
print("Final bias:", b)
```

```python
def predict(x):
    z = np.dot(w, x) + b
    return step(z)

# Test
print(predict([1, 0]))  # Spam
print(predict([0, 0]))  # Not Spam
```
