# XOR Problem using multi layer neural network (Forward propagation only)

Architecture:
2 Inputs -> 2 Hidden Neurons + 1 output


```python
import numpy as np
```

```python
X= np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y=np.array([0,1,1,0])

X,y
```

```python
w1,w2,w3,w4= 1,1,1,1
w5,w6 = 1,-2
b1=-0.5
b2 = -1.5
b3 = -0.5

```

```python
def step_function(z):
    return 1 if z>=0 else 0
```

```python
def ANN(X):
    outputs = []
    
    for x1, x2 in X:
        # Hidden layer neuron 1
        z1 = w1*x1 + w2*x2 + b1
        h1 = step_function(z1)
        
        # Hidden layer neuron 2
        z2 = w3*x1 + w4*x2 + b2
        h2 = step_function(z2)
        
        # Output neuron
        z3 = w5*h1 + w6*h2 + b3
        y_hat = step_function(z3)
        
        outputs.append(y_hat)
    
    return np.array(outputs)
```

```python
y_pred = ANN(X)
print(y_pred)
```

```python
import matplotlib.pyplot as plt
# Plot
plt.scatter(X[y_pred==0][:,0], X[y_pred==0][:,1], label="Output = 0")
plt.scatter(X[y_pred==1][:,0], X[y_pred==1][:,1], label="Output = 1")

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("ANN XOR Classification")
plt.legend()
plt.grid(True)
plt.show()
```
